```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Define the function
def function(x):
    return (x + 3) ** 2

# 2. Define the derivative
def derivative(x):
    return 2 * (x + 3)

# 3. Initialize parameters
x = 2
learning_rate = 0.1
iterations = 100

# Store values for plotting
x_values = []
y_values = []

# 4. Apply Gradient Descent
for i in range(iterations):
    gradient = derivative(x)

    x_values.append(x)
    y_values.append(function(x))

    # Update x using gradient descent
    x = x - learning_rate * gradient

    print(f"Iteration {i+1}: x = {x:.6f}, y = {function(x):.6f}")

# 5. Display final minimum
print("\nLocal Minimum:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))

# 6. Plot the function
x_range = np.linspace(-6, 4, 100)

plt.plot(x_range, function(x_range), label="y = (x+3)^2")
plt.scatter(x_values, y_values, color="red", label="Gradient Descent Steps")
plt.scatter(x, function(x), color="green", s=100, label="Minimum")

plt.title("Gradient Descent Algorithm")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
