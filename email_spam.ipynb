```python
# ==============================================================================
# ASSIGNMENT 2: EMAIL SPAM CLASSIFICATION (KNN vs. SVM)
# ==============================================================================
# ALGORITHM & OBJECTIVE:
#   Classify emails into Normal (0 - Not Spam) and Abnormal (1 - Spam).
#   1. Feature Matrix: Word frequency vectors across 3000 common dictionary words.
#   2. Preprocessing: Exclude 'Email No.' and ensure zero data leakage across splits.
#   3. K-Nearest Neighbors: Instance-based learner calculating Euclidean distance
#      to k nearest neighbors followed by majority vote.
#   4. Support Vector Machine (Linear/RBF): Maximizes margin separation hyperplane
#      between spam and legitimate word distributions.
#
# TIME & SPACE COMPLEXITIES:
#   - K-Nearest Neighbors (KNN):
#       * Training: O(1) (Lazy learner, only stores index structures).
#       * Inference: O(N_train * D) per test sample for brute-force distance.
#       * Space: O(N_train * D) to retain all training instances in RAM.
#   - Support Vector Machine (SVM):
#       * Training: O(N_train^2 * D) to O(N_train^3 * D) for quadratic programming.
#       * Inference: O(N_sv * D) where N_sv = number of support vectors.
#       * Space: O(N_sv * D) to store support vector coordinates and dual weights.
# ==============================================================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Read dataset
df = pd.read_csv('emails.csv')

# Drop non-feature column identifier
X = df.drop(columns=['Email No.', 'Prediction'], errors='ignore')
y = df['Prediction']

# Stratified split to maintain class ratio
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Feature Scaling (Crucial for distance-based algorithms like KNN & SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 1. K-Nearest Neighbors Classifier
knn = KNeighborsClassifier(n_neighbors=5, metric='euclidean', n_jobs=-1)
knn.fit(X_train_scaled, y_train)
y_pred_knn = knn.predict(X_test_scaled)

# 2. Support Vector Machine Classifier
svm = SVC(kernel='linear', C=1.0, random_state=42)
svm.fit(X_train_scaled, y_train)
y_pred_svm = svm.predict(X_test_scaled)

# Performance Evaluation
print("=" * 50)
print("K-NEAREST NEIGHBORS (KNN) EVALUATION")
print("=" * 50)
print(f"Accuracy: {accuracy_score(y_test, y_pred_knn):.4f}")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_knn))
print("Classification Report:\n", classification_report(y_test, y_pred_knn))

print("=" * 50)
print("SUPPORT VECTOR MACHINE (SVM) EVALUATION")
print("=" * 50)
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm):.4f}")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_svm))
print("Classification Report:\n", classification_report(y_test, y_pred_svm))
```
